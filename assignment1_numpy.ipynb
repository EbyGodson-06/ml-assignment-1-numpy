```python
import numpy as np

np.random.seed(42)
```

```python
##1
A = np.random.normal(
    loc=50,
    scale=12,
    size=(1000, 5)
)

print("Shape:", A.shape)
print("Data type:", A.dtype)
print("Memory size:", A.nbytes, "bytes")

A = A.astype(np.float32)

print("Data type after conversion:", A.dtype)
print("Memory size after float32:", A.nbytes, "bytes")
```

```python
##2
A = (A - A.mean(axis=0)) / A.std(axis=0)

print("Column means:")
print(np.round(A.mean(axis=0), 6))

print("\nColumn standard deviations:")
print(np.round(A.std(axis=0), 6))
```

```python
##3
def minmax(X, axis):
    minimum = X.min(axis=axis, keepdims=True)
    maximum = X.max(axis=axis, keepdims=True)

    return (X - minimum) / (maximum - minimum)
```

```python
X0 = minmax(A, axis=0)

print("Axis = 0")
print("Minimum:", X0.min(axis=0))
print("Maximum:", X0.max(axis=0))
```

```python
X1 = minmax(A, axis=1)

print("Axis = 1")
print("Minimum:", X1.min(axis=1))
print("Maximum:", X1.max(axis=1))
```

```python
##4
A = np.random.normal(
    loc=50,
    scale=12,
    size=(1000, 5)
).astype(np.float32)

num_nan = int(0.05 * A.size)

indices = np.random.choice(
    A.size,
    size=num_nan,
    replace=False
)

A.flat[indices] = np.nan

print("NaN values before replacement:", np.isnan(A).sum())
```

```python
column_means = np.nanmean(A, axis=0)

mask = np.isnan(A)

A[mask] = np.take(
    column_means,
    np.where(mask)[1]
)

print("NaN values after replacement:", np.isnan(A).sum())
```

```python
##5
column_mean = A.mean(axis=0)
column_std = A.std(axis=0)
column_median = np.median(A, axis=0)

outlier_mask = np.abs(
    A - column_mean
) > (3 * column_std)

outliers_per_column = outlier_mask.sum(axis=0)

A[outlier_mask] = np.take(
    column_median,
    np.where(outlier_mask)[1]
)

print("Number of values replaced per column:")
print(outliers_per_column)
```

```python
##6
X = A[:200]

diff = (
    X[:, np.newaxis, :]
    - X[np.newaxis, :, :]
)

distance_matrix = np.sqrt(
    np.sum(diff ** 2, axis=2)
)

print("Distance matrix shape:", distance_matrix.shape)

print(
    "Diagonal is zero:",
    np.allclose(np.diag(distance_matrix), 0)
)

print(
    "Matrix equals its transpose:",
    np.allclose(distance_matrix, distance_matrix.T)
)
```

```python
##7
D = distance_matrix.copy()

# Exclude the row itself
np.fill_diagonal(D, np.inf)

nearest_indices = np.argpartition(
    D,
    kth=4,
    axis=1
)[:, :5]

print("Nearest-neighbour array shape:")
print(nearest_indices.shape)

print("\nFirst 10 rows:")
print(nearest_indices[:10])
```

```python
##8
numbers = np.random.randint(
    0,
    100,
    size=1_000_000
)

print("Number of elements:", len(numbers))
print("Minimum value:", numbers.min())
print("Maximum value:", numbers.max())
```

```python
%%time

counts_numpy = np.bincount(
    numbers,
    minlength=100
)

print("Number of counts:", len(counts_numpy))
```

```python
%%time

counts_loop = [0] * 100

for value in numbers:
    counts_loop[value] += 1

counts_loop = np.array(counts_loop)

print("Number of counts:", len(counts_loop))
```

```python
print(
    "Both results are equal:",
    np.array_equal(counts_numpy, counts_loop)
)
```

```python
##9
from numpy.lib.stride_tricks import sliding_window_view
import pandas as pd
```

```python
data = np.random.random(10_000)

windows = sliding_window_view(
    data,
    window_shape=10
)

print("Sliding window shape:", windows.shape)
```

```python
window_means = windows.mean(axis=1)

print("Number of window means:", len(window_means))
```

```python
pandas_means = (
    pd.Series(data)
    .rolling(10)
    .mean()
    .dropna()
    .to_numpy()
)

print(
    "Results match:",
    np.allclose(window_means, pandas_means)
)
```
